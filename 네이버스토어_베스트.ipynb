```python
# 베스트 내용 수집하기
print("\n")
print("===== 곧 수집된 결과를 출력합니다 ^^ ===== ")
print("\n")

ranking2=[]         #제품의 판매순위
upce2 = []          #업체명
title2=[]           #제품 이름
price2=[]           #원래 판매가 저장
p_price2=[]         #현재 판매가 저장
discount2 = []      #할인율 저장
review2 = []        #리뷰건수
score2 = []         #평점
url2 = []           #제품 url
img_src2=[]   # 이미지 URL 저장변수
file_no = 0   # 이미지 파일 저장할 때 번호
count = 1     # 총 게시물 건수 카운트 변수

# def scroll_down(driver):      
#     driver.execute_script("window.scrollBy(0,1100);")
#     time.sleep(1)

def scroll_down(driver):
    driver.execute_script("window.scrollTo(0,document.body.scrollHeight);")
    time.sleep(2)

#for x in range(1,page_cnt + 1) :
    
# for cc in range(1,5) :
#     scroll_down(driver)

html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')

item_result = soup.find('div','homeResponsive_product_wrap__DDUTq homeResponsive_best_view__WxjKx').find_all('li')


if cnt >= len(item_result):
    cnt = len(item_result)

print('전체 검색 결과 건수 : ', len(item_result),'건')
print('실제 최종 출력 건수', cnt)
# item_result.find_all('li','productCardResponsive_product_card_responsive__yDxRB _nlog_click _nlog_impression_element')

for li in item_result :
    if cnt < count :
        break
    time.sleep(1)
    
    # 제품 이미지 다운로드 하기
    try :
        photo = li.find('div','productCardResponsive_thumbnail__ZDZh3').find('img')['src']
    except AttributeError :
        continue

    file_no += 1
    
    # photo2 = 'https:' + photo

    urllib.request.urlretrieve(photo, str(file_no)+'.jpg')
    time.sleep(0.5)

    #제품 내용 추출하기
    f = open(ff_name, 'a',encoding='UTF-8')
    f.write("-----------------------------------------------------"+"\n")
    print("-" *70)

    ranking = count
    print("1. 판매순위:",ranking)
    f.write('1. 판매순위:'+ str(ranking) + "\n")

    try :
        up = li.find('div','productCardResponsive_information__CD_1n').find('a').get_text()
    except :
        upce = '제품이름 없습니다'
        print(upce.replace("\n",""))
        f.write('2. 업체명:'+ upce + "\n")
    else :
        upce = up.replace("\n","").strip()
        print("2. 업체명:", upce.replace("\n",""))                  
        f.write('2. 업체명:'+ upce + "\n")


    try :
        t = li.find('div','productCardResponsive_information__CD_1n').find('strong').get_text()
    except :
        title = '제품이름 없습니다'
        print(title.replace("\n",""))
        f.write('3. 제품이름:'+ title + "\n")
    else :
        title = t.replace("\n","").strip()
        print("3. 제품이름:", title.replace("\n",""))                  
        f.write('3. 제품이름:'+ title + "\n")


    try :
        price = li.find('div','productCardResponsive_my_discount__Nfk7X').find('span').get_text().replace('원가','').replace('원','')
    except :
        price = '0'
        print("4. 원가:", price.replace("\n",""))
        f.write('4. 원가:'+ price + "\n")
    else :
        print("4. 원가:", price.replace("\n",""))
        f.write('4. 원가:'+ price + "\n")


    try :
        p_price = li.find('div','productCardResponsive_price_area__UZdBk').find('span','productCardResponsive_price__JeSTN').get_text().replace('원', '')
    except  :
        p_price='0'
        print('5. 현 판매가: ',p_price)
        f.write('5. 현 판매가:'+ p_price + "\n")
    else :
        print('5. 현 판매가:',p_price)
        f.write('5. 현 판매가:'+ p_price + "\n")


    try :
        discount = li.find('div','productCardResponsive_price_area__UZdBk').find('em').get_text().replace('할인율','')
    except  :
        discount = '0%'
        print("6. 할인률:", discount)
        f.write('6. 할인률:'+ discount + "\n")
    else :
        print("6. 할인률:", discount)
        f.write('6. 할인률:'+ discount + "\n")


    try :
        review = li.find('div','productCardResponsive_wrap_review__oJHZT').find('span','productCardResponsive_review__LkeRC').get_text().split(' ')[1]
    except  :
        review = '리뷰가 없습니다.'
        print("7. 리뷰수:", review)
        f.write('7. 리뷰수:'+ review + "\n")
    else :
        print("7. 리뷰수:", review)
        f.write('7. 리뷰수:'+ review + "\n")


    try :
        score = li.find('div','productCardResponsive_wrap_review__oJHZT').find('span').get_text().replace('별점','')
    except  :
        score = '0'
        print("8. 평점:", score)
        f.write('8. 평점:'+ score + "\n")
    else :
        print("8. 평점:", score)
        f.write('8. 평점:'+ score + "\n")


    try :
        url = li.find('div','productCardResponsive_inner__p8x3v').find_all('a')[1]['href']
    except  :
        url = 'URL이 없습니다.'
        print("9. url:", url)
        f.write('9. url:'+ url + "\n")
    else :
        print("9. url:", url)
        f.write('9. url:'+ url + "\n")
    
    
    print("-" *70)

    f.close( )             
    time.sleep(0.5)

    ranking2.append(ranking)
    upce2.append(upce)
    title2.append(title.replace("\n",""))
    price2.append(price.replace("\n",""))
    p_price2.append(p_price)
    discount2.append(discount)
    review2.append(review)
    score2.append(score)
    url2.append(url)

    count += 1
    
    time.sleep(1)

print('수집이 완료되었습니다.')
```

```python
#step 6. csv , xlsx 형태로 저장하기              
na_best = pd.DataFrame()
na_best['판매순위']=ranking2
na_best['업체명']=pd.Series(upce2)
na_best['제품명']=pd.Series(title2)
na_best['원가']=pd.Series(price2)
na_best['현 판매가']=pd.Series(p_price2)
na_best['할인율']=pd.Series(discount2)
na_best['리뷰 수']=pd.Series(review2)
na_best['평점']=pd.Series(score2)
na_best['URL']=pd.Series(url2)

# csv 형태로 저장하기
na_best.to_csv(fc_name,encoding="utf-8-sig",index=False)

# 엑셀 형태로 저장하기
na_best.to_excel(fx_name ,index=False)

e_time = time.time( )
t_time = e_time - s_time

count -= 1
print("\n")
print("=" *100)
print("1.요청된 총 %s 건의 리뷰 중에서 실제 크롤링 된 리뷰 수는 %s 건입니다" %(cnt,count))
print("2.총 소요시간은 %s 초 입니다 " %round(t_time,1))
print("3.파일 저장 완료: txt 파일명 : %s " %ff_name)
print("4.파일 저장 완료: csv 파일명 : %s " %fc_name)
print("5.파일 저장 완료: xlsx 파일명 : %s " %fx_name)
print("=" *100)
```

```python
#DB연동을 위한, 코드
import excelrd, pymysql

xlsx = excelrd.open_workbook(fx_name)
sheet = xlsx.sheet_by_name('Sheet1')

mysql = pymysql.connect(host='192.168.25.17', port=3306,
                        user='my_user',
                        passwd='mysql',
                        db='my_testdb',
                        charset='utf8')

cursor = mysql.cursor()

query_drop = "DROP TABLE IF EXISTS n_best"
cursor.execute(query_drop)

#테이블 생성
query_1 = '''
CREATE TABLE n_best(
ranking TEXT,
upce TEXT,
title TEXT,
price TEXT,
p_price TEXT,
discount TEXT,
review TEXT,
score TEXT,
url TEXT
);
'''
cursor.execute(query_1)

# DB 삽입
query_2 = '''INSERT INTO n_best(
ranking, upce, title, price, p_price, discount, review, score, url) VALUES(%s, %s, %s, %s, %s, %s, %s, %s, %s)'''

for r in range(1, sheet.nrows):
    ranking    = sheet.cell(r, 0).value
    upce = sheet.cell(r, 1).value
    title = sheet.cell(r, 2).value
    price = sheet.cell(r, 3).value  
    p_price = sheet.cell(r, 4).value  
    discount  = sheet.cell(r, 5).value  
    review = sheet.cell(r, 6).value  
    score = sheet.cell(r, 7).value  
    url = sheet.cell(r, 8).value 

    values = (ranking, upce, title, price, p_price, discount, review, score, url)
    cursor.execute(query_2, values)

mysql.commit()  # 저장
cursor.close()  # 닫기
mysql.close()   # 닫기
```
